```python
from pathlib import Path

# Carpeta principal donde quieres crear la estructura
base_dir = Path(".")

# Carpetas principales y subcarpetas
main_folders = ["F200", "F100", "F50"]
subfolders = ["100", "75", "50", "25"]

# Crear estructura
for main in main_folders:
    for sub in subfolders:
        folder = base_dir / main / sub
        folder.mkdir(parents=True, exist_ok=True)

print("Estructura de carpetas creada.")
```
